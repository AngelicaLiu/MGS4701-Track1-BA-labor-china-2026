```python
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix, cohen_kappa_score

# 1. 读取数据（确保这三个文件都在 notebooks 文件夹的上一级或同级）
machine_df = pd.read_csv('../data/documentation/machine_extraction_results_A2.csv')
gold_150 = pd.read_csv('../data/documentation/Goldset_Coder_B_150.csv')

# 2. 将机器结果与 B 同学的 150 条人工结果合并（以 record_id 为基准）
merged_150 = pd.merge(machine_df, gold_150[['record_id', 'gold_ai_flag']], on='record_id', how='inner')

# 3. 提取真实值和预测值，计算 F1
y_true = merged_150['gold_ai_flag']
y_pred = merged_150['has_AI_Skills']

print("=== Machine Extraction vs Human Gold Set (150 items) ===")
print(f"Total Sample: {len(merged_150)}")
print(f"Precision: {precision_score(y_true, y_pred):.3f}")
print(f"Recall: {recall_score(y_true, y_pred):.3f}")
print(f"F1-Score: {f1_score(y_true, y_pred):.3f}")
print("Confusion Matrix:")
print(confusion_matrix(y_true, y_pred))

# 4. 计算 B 和 C 的一致性 (Kappa)
# 假设 C 同学交的表叫 Goldset_Coder_C_50.csv
gold_c_50 = pd.read_csv('../data/documentation/Goldset_Coder_C_50.csv')

# 取前 50 条进行合并
merged_50 = pd.merge(gold_150.head(50), gold_c_50, on='record_id', suffixes=('_B', '_C'))

print("\n=== Inter-coder Agreement (B vs C, 50 items) ===")
print(f"Total Sample: {len(merged_50)}")
kappa = cohen_kappa_score(merged_50['gold_ai_flag_B'], merged_50['gold_ai_flag_C'])
print(f"Cohen's Kappa: {kappa:.3f}")
print("Confusion Matrix:")
print(confusion_matrix(merged_50['gold_ai_flag_B'], merged_50['gold_ai_flag_C']))
```
